```python
import numpy as np
import cv2 as cv
import os
import glob
import matplotlib.pyplot as plt
from numpy.random import uniform
import pdb
```

```python
def find_color_values_using_trackbar(frame):

    frame_hsv = cv.cvtColor(frame, cv.COLOR_BGR2HSV)
 
    def nothing(x):
        pass

    cv.namedWindow("Trackbar") 
    cv.createTrackbar("LH", "Trackbar", 0, 255, nothing)
    cv.createTrackbar("LS", "Trackbar", 0, 255, nothing)
    cv.createTrackbar("LV", "Trackbar", 0, 255, nothing)
    cv.createTrackbar("UH", "Trackbar", 255, 255, nothing)
    cv.createTrackbar("US", "Trackbar", 255, 255, nothing)
    cv.createTrackbar("UV", "Trackbar", 255, 255, nothing)
    
    
    while True:

        l_h = cv.getTrackbarPos("LH", "Trackbar")
        l_s = cv.getTrackbarPos("LS", "Trackbar")
        l_v = cv.getTrackbarPos("LV", "Trackbar")
        u_h = cv.getTrackbarPos("UH", "Trackbar")
        u_s = cv.getTrackbarPos("US", "Trackbar")
        u_v = cv.getTrackbarPos("UV", "Trackbar")


        l = np.array([l_h, l_s, l_v])
        u = np.array([u_h, u_s, u_v])
        mask_table_hsv = cv.inRange(frame_hsv, l, u)        

        res = cv.bitwise_and(frame, frame, mask=mask_table_hsv)    
        cv.imshow("Frame", frame)
        cv.imshow("Mask", mask_table_hsv)
        cv.imshow("Res", res)

        if cv.waitKey(25) & 0xFF == ord('q'):
                break
    cv.destroyAllWindows()
```

```python
import cv2 as cv
import numpy as np

# === 1. CHANGE THIS TO YOUR IMAGE PATH ===
IMG_PATH = "C:/Users/drago/Downloads/CAVA_2025_2026_NEW/CAVA-2025-TEMA1/data/imagini_auxiliare/08.jpg"

def nothing(x):
    pass

# Load image
img = cv.imread(IMG_PATH)

if img is None:
    raise Exception(f"Failed to load image from {IMG_PATH}")

# Optional: resize if it's huge
scale = 0.7
img = cv.resize(img, None, fx=scale, fy=scale, interpolation=cv.INTER_AREA)

# Convert once to HSV
hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)

# === 2. CREATE TRACKBARS ===
cv.namedWindow("Trackbars")
cv.createTrackbar("LH", "Trackbars", 0,   179, nothing)
cv.createTrackbar("LS", "Trackbars", 0,   255, nothing)
cv.createTrackbar("LV", "Trackbars", 0,   255, nothing)
cv.createTrackbar("UH", "Trackbars", 179, 179, nothing)
cv.createTrackbar("US", "Trackbars", 255, 255, nothing)
cv.createTrackbar("UV", "Trackbars", 255, 255, nothing)

print("Controls:")
print(" - Move sliders until ONLY the color you want is white in the mask.")
print(" - Press 's' to print the current HSV range.")
print(" - Press 'q' to quit.")

while True:
    # Read trackbar values
    lh = cv.getTrackbarPos("LH", "Trackbars")
    ls = cv.getTrackbarPos("LS", "Trackbars")
    lv = cv.getTrackbarPos("LV", "Trackbars")
    uh = cv.getTrackbarPos("UH", "Trackbars")
    us = cv.getTrackbarPos("US", "Trackbars")
    uv = cv.getTrackbarPos("UV", "Trackbars")

    lower = np.array([lh, ls, lv])
    upper = np.array([uh, us, uv])

    # Apply mask
    mask = cv.inRange(hsv, lower, upper)
    result = cv.bitwise_and(img, img, mask=mask)

    # Show original + mask
    cv.imshow("image", img)
    cv.imshow("mask", mask)
    cv.imshow("result", result)

    key = cv.waitKey(1) & 0xFF
    if key == ord('q'):
        break
    if key == ord('s'):
        print(f"Selected HSV range: lower={tuple(lower)}, upper={tuple(upper)}")

cv.destroyAllWindows()
```
